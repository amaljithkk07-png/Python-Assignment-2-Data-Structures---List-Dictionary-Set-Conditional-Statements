```python
age_list = [24, 25, 26, 27, 28]
print(age_list)
name_list = ["Amal", "Arun", "Anu", "Meera", "Riya"]
print(name_list)
```

```python
# 2. List Operations / Modifications

name_list.append("Yazhini")
print("After append:", name_list)

age_list.insert(2, 30)
print("After insert:", age_list)

name_list.remove("Yazhini")
print("After remove:", name_list)

age_list.pop()
print("After pop:", age_list)

age_list.extend([29, 30, 26])
print("After extend:", age_list)

age_list.sort(reverse=True)
print("Descending order:", age_list)

print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# 3. Accessing List Elements

# a. Print the first element of name_list
print("First element:", name_list[0])

# b. Print the last element of name_list
print("Last element:", name_list[-1])

# c. Print elements from index 2 to index 4
print("Elements from index 2 to 4:", name_list[2:5])

# d. Print elements in reverse order
print("Reverse list:", name_list[::-1])
```

```python

# 4. Dictionary (Creation, Modification and Access)

# a. Create a dictionary of five students and their marks
student_marks = {
    "Amal": 75,
    "Arun": 88,
    "Anu": 92,
    "Meera": 68,
    "Riya": 95
}

print("Student Marks:", student_marks)

# b. Access and print the mark of a specific student
print("Amal's Mark:", student_marks["Amal"])

# c. Add a new student Janani with mark 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update an older student's mark to 82
student_marks["Amal"] = 82
print("After updating Amal's mark:", student_marks)

# e. Print keys, values, and key-value pairs
print("All Keys:", student_marks.keys())
print("All Values:", student_marks.values())
print("All Items:", student_marks.items())
```

```python

# 5. Sets (Operations)

# a. Create a set with the given values
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("My Set:", my_set)

# Explanation:
print("Duplicate values are removed automatically in a set.")

# b. Attempt to change an element of the set
try:
    my_set[4] = 's'
except TypeError as e:
    print("Error:", e)
    print("Explanation: Sets are unindexed and do not support item assignment.")

# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)

# d. Compute union and intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

```python

# 6. Performance Category Program

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent performance! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
